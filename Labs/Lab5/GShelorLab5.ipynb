```python
### importing libraries
import arcpy
import os

### set arcpy to overwrite existing files
arcpy.env.overwriteOutput = True

### setting workspace
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5/Lab5_Data/Lab5_Routes"

### setting default output folder
outWorkspace = "C:/Users/gshelor/Documents/Lab5/Lab5Output/RouteLines"
```

```python
### converting points to lines
for i in arcpy.ListFeatureClasses():
    lineroutes = os.path.join(outWorkspace, i)
    arcpy.PointsToLine_management(i, lineroutes)
    ### Report progress
    print(arcpy.GetMessages())
```

```python
##### reprojecting line feature classes #####
### setting new workspaces
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5/Lab5Output/RouteLines"
outWorkspace2 = "C:/Users/gshelor/Documents/Lab5/Lab5Output/ProjLines"
for i in arcpy.ListFeatureClasses():
    projlineroutes = os.path.join(outWorkspace2, i)
    ### set new CRS (UTM Zone 11N)
    newCS = arcpy.SpatialReference(26911)
        
    ### reproject to appropriate CRS
    arcpy.Project_management(i, projlineroutes, newCS)
        
    ### Report progress
    print(arcpy.GetMessages())
```

```python
##### Adding a field to reprojected lines #####
### ArcPro's documentation says the arcpy.AddGeometryAttributes_management tool is deprecated, at least for ArcPro 3.3
### setting new workspaces
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5/Lab5Output/ProjLines"
### don't think I need this actually
# outWorkspace3 = "C:/Users/gshelor/Documents/Lab5/Lab5Output/ProjLengthLines"

for i in arcpy.ListFeatureClasses():
    # linelengthroutes = os.path.join(outWorkspace3, i)
        
    ### calculating length of lines in miles
    arcpy.management.CalculateGeometryAttributes(i, [["line_len","LENGTH"]], "MILES_INT")
        
    ### Report progress
    print(arcpy.GetMessages())
```

```python
##### buffering lines by 200 feet #####
### setting new workspaces
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5/Lab5Output/ProjLines"
outWorkspace4 = "C:/Users/gshelor/Documents/Lab5/Lab5Output/LineBuffers"

for i in arcpy.ListFeatureClasses():
    linebufferroutes = os.path.join(outWorkspace4, i)
    
    ### creating buffers
    arcpy.analysis.PairwiseBuffer(i, linebufferroutes, "200 Feet", "ALL")
    
    ### Report progress
    print(arcpy.GetMessages())
```

```python
##### Loading bike lane shapefiles and intersecting them with bike route buffers #####
### setting new workspaces
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5"
outWorkspace5 = "C:/Users/gshelor/Documents/Lab5/Lab5Output/BikeLaneIntersects"

### reading in bike lanes
bike_lanes = "/Lab5_Data/RenoBikeLanes.shp"

### resetting workspace for the buffers
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5/Lab5Output/LineBuffers"



### performing intersect
for i in arcpy.ListFeatureClasses():
    arcpy.analysis.PairwiseIntersect([bike_lanes, i], os.path.join(outWorkspace5, i))
    ### Report progress
    print(arcpy.GetMessages())

```

```python
##### Adding a field to intersect output to determine length of interset feature in miles #####
### setting new workspaces
arcpy.env.workspace = "C:/Users/gshelor/Documents/Lab5/Lab5Output/BikeLaneIntersects"
# outWorkspace6 = "C:/Users/gshelor/Documents/Lab5/Lab5Output/BikeLaneIntersects"

for i in arcpy.ListFeatureClasses():
    ### calculating length of intersects in miles
    arcpy.management.CalculateGeometryAttributes(i, [["line_len","LENGTH"]], "MILES_INT")
    ### Report progress
    print(arcpy.GetMessages())
```
